### Quiz Game!
 "Which continent does china located?",
    "What is the name of kenya president?",
    "What is the capital of Somalia?",
    "What is the second largest continent in the world?"



```python
questions = [
    "What is the largest city in the world?",
    "Which continent is China located in?",
    "Who is the president of Kenya?",
    "What is the capital of Somalia?",
    "What is the second largest continent in the world?"
]

choices = [
    ["A. Jakarta", "B. New York", "C. Tokyo", "D. New Delhi"],

    ["A. Africa", "B. Asia", "C. Europe", "D. Australia"],

    ["A. William Ruto", "B. Uhuru Kenyatta", "C. Raila Odinga", "D. Mwai Kibaki"],

    ["A. Hargeisa", "B. Mogadishu", "C. Nairobi", "D. Kismayo"],

    ["A. Asia", "B. Europe", "C. Africa", "D. Australia"]
]

answers = ["a", "b", "c", "b", "c"]
score = 0

for i in range(len(questions)):
    print(questions[i])

    for choice in choices[i]:
        print(choice)

    answer = input("Enter your answer (A/B/C/D): ").lower()
    print("Your answer is:", answer)

    if answer == answers[i]:
        print("Correct! and you are genius")
        score += 20

    else:
            print("Wrong answer and you are idiot hhh")
   

print("Your final score is:", score, "out of", len(questions) "questions")


if score >= 80:
    print("Excelent")

elif score >= 50:
        print("Good")


else:
            print("Try again")


```
